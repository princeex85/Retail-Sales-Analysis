# Retail Sales Analysis

## Notebook 04 – Feature Engineering

### Objective

The purpose of this notebook is to create additional features from the cleaned dataset to support deeper analysis and improve dashboard development.

The engineered features will enhance reporting, trend analysis, and business decision-making.

```python
# =====================================
# Import Libraries
# =====================================

import pandas as pd
import numpy as np
```

```python
# =====================================
# Load Cleaned Dataset
# =====================================

df = pd.read_csv("../Data/processed/Cleaned_Superstore.csv")
```

## Feature 1: Order Year

Extract the year from the Order Date column to support yearly trend analysis.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Order Year"] = df["Order Date"].dt.year

df[["Order Date", "Order Year"]].head()
```

## Feature 2: Order Month

Extract the month number from the Order Date for monthly analysis.

```python
df["Order Month"] = df["Order Date"].dt.month

df[["Order Date", "Order Month"]].head()
```

## Feature 3: Month Name

Create the month name to improve readability in reports and dashboards.

```python
df["Month Name"] = df["Order Date"].dt.month_name()

df[["Order Date", "Month Name"]].head()
```

## Feature 4: Quarter

Identify the quarter of each order for quarterly business analysis.

```python
df["Quarter"] = df["Order Date"].dt.quarter

df[["Order Date", "Quarter"]].head()
```

## Feature 5: Shipping Duration

Calculate the number of days taken to ship each order.

```python
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

df[["Order Date", "Ship Date", "Shipping Days"]].head()
```

## Feature 6: Profit Margin

Calculate the profit margin as a percentage of sales.

```python
df["Profit Margin (%)"] = (
    df["Profit"] / df["Sales"]
) * 100

df[["Sales", "Profit", "Profit Margin (%)"]].head()
```

## Feature 7: Sales Category

Classify orders into Low, Medium, and High sales categories based on sales value.

```python
df["Sales Category"] = pd.cut(
    df["Sales"],
    bins=[0, 100, 500, df["Sales"].max()],
    labels=["Low", "Medium", "High"]
)

df[["Sales", "Sales Category"]].head()
```

## Feature 8: Discount Category

Group discount percentages into meaningful business categories.

```python
df["Discount Category"] = pd.cut(
    df["Discount"],
    bins=[-0.01, 0, 0.20, 0.50, 1],
    labels=[
        "No Discount",
        "Low Discount",
        "Medium Discount",
        "High Discount"
    ]
)

df[["Discount", "Discount Category"]].head()
```

## Feature 9: Profit Category

Categorize transactions based on profit values.

```python
df["Profit Category"] = pd.cut(
    df["Profit"],
    bins=[
        df["Profit"].min(),
        0,
        500,
        df["Profit"].max()
    ],
    labels=[
        "Loss",
        "Low Profit",
        "High Profit"
    ]
)

df[["Profit", "Profit Category"]].head()
```

## Final Dataset Validation

Review the newly created features before exporting the dataset.

Export the dataset for use in Power BI and future analysis.
