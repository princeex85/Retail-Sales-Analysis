## Notebook 02 – Data Cleaning

### Objective
This notebook focuses on identifying and resolving data quality issues before exploratory data analysis. The cleaning process includes:

- Missing values
- Duplicate records
- Incorrect data types
- Invalid values
- Standardizing text
- Final cleaned dataset

```python
# =====================================
# Import Libraries
# =====================================

import pandas as pd
import numpy as np
```

```python
# =====================================
# Load Dataset
# =====================================


df = pd.read_excel("../Data/Superstore.csv.xlsx")
```

## 1. Check Missing Values

Missing values can affect calculations and visualizations.
The first step is to identify whether any column contains null values.

```python
df.isnull().sum()
```

```python
print("Duplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates()
```

### Observation

No missing values were found in the dataset.

No further action is required.

## 3. Validate Data Types

Ensuring that every column has the appropriate data type is essential for accurate analysis and calculations.

```python
df.dtypes
```

```python
# Convert date columns to datetime format

df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
# Verify updated data types

df.dtypes
```

### Observation

The date columns were converted to the appropriate datetime format.

This enables accurate time-based analysis in later stages of the project.

## 4. Standardize Text Values

Text columns may contain leading or trailing spaces, which can create inconsistencies during grouping and filtering.

This step removes unnecessary whitespace from all text columns.

```python
# Identify text columns

text_columns = df.select_dtypes(include="object").columns

text_columns
```

```python
# Remove leading and trailing spaces

for column in text_columns:
    df[column] = df[column].str.strip()
```

### Observation

Leading and trailing spaces were removed from all text columns to ensure consistency.

## 5. Validate Numerical Columns

This step reviews the numerical columns to identify invalid or unrealistic values that could affect analysis.

```python
df.describe()
```

### Observation

The numerical columns were reviewed for invalid values.

No abnormal values requiring correction were identified.

## 6. Final Dataset Validation

Performing a final review of the cleaned dataset before saving it for Exploratory Data Analysis.

```python
df.info()
```

```python
df.head()
```

## 7. Save Clean Dataset

The cleaned dataset is exported for use in the Exploratory Data Analysis (EDA) notebook.

```python
df.to_csv("../Data/Cleaned_Superstore.csv", index=False)

print("Cleaned dataset saved successfully.")
```

# Cleaning Summary

The following data cleaning tasks were completed successfully:

- Verified missing values
- Removed duplicate records
- Validated data types
- Converted date columns to datetime format
- Standardized text values
- Reviewed numerical columns
- Saved the cleaned dataset

The dataset is now ready for Exploratory Data Analysis (EDA).
